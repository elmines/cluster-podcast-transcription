```python
from transformers import AutoTokenizer
```

```python
data_dir = "resegment_out/"
```

```python
tokenizer_names = [
    "google-bert/bert-base-cased",
    "answerdotai/ModernBERT-base",
    "meta-llama/Llama-3.3-70B-Instruct",
    "google/gemma-4-E2B-it",
    "openai/gpt-oss-20b"
]
tokeninzers = {k:AutoTokenizer.from_pretrained(k) for k in tokenizer_names}
```

```python
[type(v) for v in tokeninzers.values()]
```

```python
import glob
import os
import csv
from tqdm import tqdm
len_samples = {k:[] for k in tokenizer_names}

# Sample one episode from each show
show_dirs = [d_path for d_path in glob.glob(os.path.join(data_dir, "*")) if os.path.isdir(d_path)]
for show_dir in tqdm(show_dirs, desc="Processing shows"):
    print(show_dir)
    for csv_path in glob.glob(os.path.join(show_dir, "*.csv")):
        print('\t', csv_path)
        with open(csv_path) as r:
            texts = [row['text'] for row in csv.DictReader(r)]
        if not texts:
            continue
        for tok_name, tok in tokeninzers.items():
            tokenized = tok(texts)
            len_samples[tok_name].extend(len(sample) for sample in tokenized['input_ids'])
        # We just want to find one episode from the show with text
        break
    else:
        print(f"Failed to find text for {show_dir}")
```

```python
import plotly.graph_objects as go
import numpy as np
```

```python
out_dir = "token_counts"
os.makedirs(out_dir, exist_ok=True)
for tokenizer_name, counts in len_samples.items():
    fig = go.Figure(data=[go.Histogram(x=np.array(counts))])
    fig.update_layout(title=f"{tokenizer_name} Token Counts")
    fig.write_image(os.path.join(out_dir, tokenizer_name.replace('/', '--') + ".png"))
```
