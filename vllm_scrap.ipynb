```python
import pandas as pd
import re
from vllm import LLM, SamplingParams
```

```python
# model_name = "meta-llama/Llama-3.1-8B-Instruct"
# model_name = "meta-llama/Llama-3.2-3B-Instruct"
# model_name = "google/gemma-4-E4B-it"
# model_name = "meta-llama/Llama-4-Scout-17B-16E-Instruct"
# model_name = "meta-llama/Llama-3.3-70B-Instruct"
model_name = "openai/gpt-oss-120b"
max_num_seqs = 8
max_model_len = 2048
max_num_batched_tokens = max_num_seqs * max_model_len
llm = LLM(model=model_name,
        max_num_seqs=max_num_seqs,
        max_model_len=max_model_len,
        max_num_batched_tokens=max_num_batched_tokens,
        gpu_memory_utilization=0.95)
```

```python
from hot_topic.utils import read_transcription_text
```

```python
df = pd.read_csv("meta-llama--Llama-3.2-3B-Instruct-out/topic_quotes.csv", index_col='topic')
df.fillna('', inplace=True)
```

```python
left_context_size = 4096
right_context_size = 4096
joined_texts = []
for _, row in df.iterrows():
    quote_text = row['episode_quote']
    text = read_transcription_text(row['episode_file'])
    index = text.index(quote_text)
    left_context = text[max(0, index - left_context_size):index]
    right_context = text[index + len(text):]
    joined_texts.append(left_context + f"<document>{quote_text}</document>" + right_context)
df['joined'] = joined_texts
```

```python
sampling_params = SamplingParams(
    temperature=0.0,
    max_tokens=1024
)
```

```python
prompt_template = \
r"""
We're scrubbing advertisements, promotional messages, public service announcements, and music from podcast transcripts.
Determine whether the content in <document> tags falls in any of those categories:
{Document}
"""

system_string = \
rf"""
You are a helpful assistant. After any explanations, give a one-word yes or no answer.
"""
```

```python
yes_patt = re.compile(r'[^\n\r]*(y(?:es)?)[^\n\r]*$', flags=re.I)
```

```python
convos = [
    [
        {"role": "system", "content": system_string},
        {"role": "user", "content": prompt_template.format(Document=quote)}
    ]
    for quote in df.joined
]
```

```python
from functools import partial
from tqdm import tqdm
outputs = llm.chat(convos, sampling_params=sampling_params, use_tqdm=partial(tqdm, desc='chat convos'))
```

```python
responses = [o.outputs[0].text for o in outputs]
sentinels = ["it is an advertisement"]
sure_ad = [bool(yes_patt.search(res)) or any(s in res for s in sentinels) for res in responses]
```

```python
df['sure_is_ad'] = sure_ad
df['ad_explanation'] = responses
df.drop(columns="joined", inplace=False).to_csv("filtered_gpt120.csv")
```
